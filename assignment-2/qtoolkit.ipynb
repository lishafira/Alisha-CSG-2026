```python
import numpy as np

def ket(amps):
    return np.array(amps, dtype=complex).reshape(-1, 1)

def dag(M):
    return np.conjugate(np.array(M)).T

def inner(a, b):
    return np.dot(dag(a), b)[0, 0]

def normalize(psi):
    norm = np.sqrt(np.real(inner(psi, psi)))
    if norm == 0:
        raise ValueError("Zero vector cannot be normalized.")
    return psi / norm

def is_hermitian(M, tol=1e-9):
    M = np.array(M, dtype=complex)
    return np.allclose(M, dag(M), atol=tol)

def is_unitary(M, tol=1e-9):
    M = np.array(M, dtype=complex)
    I = np.eye(M.shape[0])
    return np.allclose(np.dot(dag(M), M), I, atol=tol)

def probabilities(psi, basis):
    probs = []
    for b in basis:
        amp = inner(b, psi)
        probs.append(float(np.real(amp * np.conjugate(amp))))
    return probs

def measure(psi, basis, shots=1000):
    probs = probabilities(psi, basis)
    outcomes = np.random.choice(len(basis), size=shots, p=probs)
    counts = {}
    for i in range(len(basis)):
        counts[f"ket_{i}"] = int(np.sum(outcomes == i))
    return counts

assert np.allclose(ket([1, 1]), np.array([[1], [1]], dtype=complex))
assert is_hermitian([[2, 1-1j], [1+1j, 3]])
assert is_unitary([[1, 0], [0, 1j]])
print("Semua fungsi toolkit berhasil diuji!")

```

```python
def random_state():
    theta = np.random.uniform(0, np.pi)
    phi = np.random.uniform(0, 2 * np.pi)
    psi = np.array([
        [np.cos(theta / 2)],
        [np.exp(1j * phi) * np.sin(theta / 2)]
    ], dtype=complex)
    return psi

X = np.array([[0, 1], [1, 0]], dtype=complex)
Y = np.array([[0, -1j], [1j, 0]], dtype=complex)
Z = np.array([[1, 0], [0, -1]], dtype=complex)
H = (1 / np.sqrt(2)) * np.array([[1, 1], [1, -1]], dtype=complex)
S_gate = np.array([[1, 0], [0, 1j]], dtype=complex)
operators = {'X': X, 'Y': Y, 'Z': Z, 'H': H, 'S': S_gate}
num_states = 1000
all_valid = True

for _ in range(num_states):
    psi = random_state()
    for name, U in operators.items():
        u_psi = np.dot(U, psi)
        norm_val = np.sqrt(np.real(inner(u_psi, u_psi)))
        if not np.isclose(norm_val, 1.0, atol=1e-7):
            all_valid = False

print(f"Verifikasi 1000 state acak untuk U in {list(operators.keys())}: {'BERHASIL (Semua norm = 1)' if all_valid else 'GAGAL'}")
```
